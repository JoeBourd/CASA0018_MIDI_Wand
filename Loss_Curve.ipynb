```python
#import regular expressions
import re

train_loss = []
val_loss = []

#employ search rules to obtain values from log
pattern = r'loss[:]\s*([0-9.]+).*val_loss[:]\s*([0-9.]+)'

#read file using search rules and append values to list
with open('logs47273963.txt', 'r') as f:
    for line in f:
        match = re.search(pattern, line)
        if match:
            train_loss.append(float(match.group(1)))
            val_loss.append(float(match.group(2)))

#plot values
plt.figure()
plt.plot(train_loss, label='Training Loss')
plt.plot(val_loss, label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.title('Training vs Validation Loss')
plt.show()
```
